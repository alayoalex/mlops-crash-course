```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Reference (training) data
ref = pd.DataFrame({
    "age": np.random.normal(30, 4, 1000),
    "income": np.random.normal(60000, 10000, 1000),
    "transactions": np.random.normal(12, 3, 1000)
})

# Current production data with mild drift
curr = pd.DataFrame({
    "age": np.random.normal(30, 4, 1000),                # no drift
    "income": np.random.normal(70000, 15000, 1000),      # mean + variance drift
    "transactions": np.random.normal(15, 5, 1000)        # moderate drift
})
```

### Run a Drift Report

```python
from evidently import Report
from evidently.presets import DataDriftPreset

report = Report(
    metrics=[
        DataDriftPreset()
    ],
include_tests=True)

eval = report.run(
    reference_data=ref,
    current_data=curr
)

# eval.save_html("drift_report.html")

eval
```

### Data Summary/Quality Report

```python
from evidently import Report
from evidently.presets import DataSummaryPreset

ds_report = Report(
    metrics=[
        DataSummaryPreset()
    ],
include_tests=True)

ds = ds_report.run(
    reference_data=ref,
    current_data=curr
)

# ds.save_html("data_summary.html")

ds
```
